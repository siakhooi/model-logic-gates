# Data Preparation (Pandas)

This notebook prepares the logic gates dataset for model training.

## 1. Import Libraries

```python
import pandas as pd
```

## 2. Load Data

```python
# Load the dataset
df = pd.read_csv('../data/raw/input.csv')
print(f"Dataset shape: {df.shape}")
df
```

## 3. Explore Data

### Data Types (df.dtypes)

```python
df.dtypes
```

### Data Info (df.info())

```python
df.info()
```

### Check for Missing Value

```python
# Check for missing values
print("Missing Values:")
print(df.isnull().sum())

print(f"\nTotal missing values: {df.isnull().sum().sum()}")
```

### Statistical summary

```python
df.describe()
```

## 4. Transform Data

### 4.1 Convert fields to float

```python
# Convert columns A, B, Result to float
df[['A', 'B', 'Result']] = df[['A', 'B', 'Result']].astype(float)

# Convert Operator column to uppercase string
df['Operator'] = df['Operator'].astype(str).str.upper()
```

### 4.2 One-Hot Encoding for Operator

```python
# One-hot encode the Operator column
df = pd.get_dummies(df, columns=['Operator'], prefix='Op')

# Display the encoded columns
print("Columns after one-hot encoding:")
print(df.columns.tolist())
df
```

### 4.3 Reorder Columns (Result as last column)

```python
# Move 'Result' column to the end
cols = [col for col in df.columns if col != 'Result'] + ['Result']
df = df[cols]

print("Reordered columns:")
print(df.columns.tolist())
df
```

## 6. Save Prepared Data

### prepare output directory

```python
import os
output_dir = '../target/prepared'
os.makedirs(output_dir, exist_ok=True)
```

### Save as CSV

```python
csv_path = os.path.join(output_dir, 'prepared_data.csv')
df.to_csv(csv_path, index=False)
print(f"✓ Saved CSV: {csv_path}")
```

### Save as Parquet

```python
parquet_path = os.path.join(output_dir, 'prepared_data.parquet')
df.to_parquet(parquet_path, index=False)
print(f"✓ Saved Parquet: {parquet_path}")
```

### Verify saved files

```python
print("Saved files:")
for f in os.listdir(output_dir):
    filepath = os.path.join(output_dir, f)
    size = os.path.getsize(filepath)
    print(f"  - {f} ({size} bytes)")
```
