# Model Inference

Load the Logic Gates model and make predictions using inference input data.

```python
import pandas as pd
import tensorflow as tf
```

## Load the input data

```python
input_df = pd.read_csv('../inference/input/input.csv')
```

```python
input_df
```

## Load the Keras model

```python
model = tf.keras.models.load_model('../models/logic_gates_model.keras')
model.summary()
```

## Prepare input data for prediction

```python
# Convert columns A, B, Result to float
input_df[['A', 'B']] = input_df[['A', 'B']].astype(float)

# Convert Operator column to uppercase string
input_df['Operator'] = input_df['Operator'].astype(str).str.upper()
```

```python
# One-hot encode the Operator column
input_df = pd.get_dummies(input_df, columns=['Operator'], prefix='Op')

# Display the encoded columns
print("Columns after one-hot encoding:")
print(input_df.columns.tolist())
```

```python
input_df
```

```python
X = input_df[['A', 'B', 'Op_AND', 'Op_NAND', 'Op_NOR', 'Op_OR', 'Op_XNOR', 'Op_XOR']]
```

## Predictions

```python
predictions = model.predict(X)
```

## Process Results

```python

# Add predictions to the dataframe
input_df['Prediction'] = predictions

# Convert probabilities to class labels (0 or 1)
input_df['Predicted_Class'] = (predictions > 0.5).astype(int)
```

## Predictions

```python
input_df
```

## Save Predictions

```python
import os
output_dir = '../inference/output'
os.makedirs(output_dir, exist_ok=True)
```

### Save to CSV File

```python
output_csv_path = os.path.join (output_dir, 'output.csv')
input_df.to_csv(output_csv_path, index=False)
```

### Save to JSON File

```python
output_json_path = os.path.join (output_dir, 'output.json')

input_df.to_json(output_json_path, orient='records', indent=2)
```

### Summary

```python
print(f"Predictions saved to:\n  - {output_csv_path}\n  - {output_json_path}")
```
