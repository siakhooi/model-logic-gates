## Import Libraries

```python
import pandas as pd
import numpy as np
import pickle
import os
```

## Load the Pickle Model

```python
model_path = '../target/models/logic_gates_model_sklearn.pkl'
with open(model_path, 'rb') as f:
    model = pickle.load(f)
print(f'Model loaded from: {model_path}')
```

## Load Inference Input Data

```python
input_path = '../inference/input/input.csv'
df_input = pd.read_csv(input_path)
print(f'Input data shape: {df_input.shape}')
df_input.head()
```

```python
# Convert columns A, B, Result to float
df_input[['A', 'B']] = df_input[['A', 'B']].astype(float)

# Convert Operator column to uppercase string
df_input['Operator'] = df_input['Operator'].astype(str).str.upper()
```

```python
# One-hot encode the Operator column
df_input = pd.get_dummies(df_input, columns=['Operator'], prefix='Op')

# Display the encoded columns
print("Columns after one-hot encoding:")
print(df_input.columns.tolist())
```

```python
df_input
```

## Prepare Features for Inference

```python
# Adjust feature columns as needed to match training
feature_cols = ['A', 'B', 'Op_AND', 'Op_NAND', 'Op_NOR', 'Op_OR', 'Op_XNOR', 'Op_XOR']
X_infer = df_input[feature_cols].values.astype(np.float32)
print(f'Inference features shape: {X_infer.shape}')
```

## Run Inference

```python
y_pred = model.predict(X_infer)
print('Predictions:', y_pred)
```

## Save Predictions to CSV

### Prepare Inference directory

```python
inference_dir = "../target/inference"
os.makedirs(inference_dir, exist_ok=True)
```

```python
output_csv = os.path.join(inference_dir, 'output_sklearn.csv')
df_output = df_input.copy()
df_output['Prediction'] = y_pred
df_output.to_csv(output_csv, index=False)
print(f'Predictions saved to: {output_csv}')
```
